```python
#a.
import numpy as np
from scipy.stats import t

x = np.array([32, 33, 34, 35, 36, 37, 38, 39, 40])
f = np.array([2, 3, 15, 26, 28, 6, 8, 8, 4])
n = f.sum()

# Trung bình mẫu
tb = (x * f).sum() / n

# Độ lệch chuẩn mẫu
dlc = np.sqrt(((x**2 * f).sum() - n * tb**2) / (n - 1))

# Khoảng tin cậy 95%
dotincay = 0.95
alpha = 1 - dotincay
saiso = t.isf(alpha/2, df=n-1) * dlc / np.sqrt(n)

print('n =', n)
print('Trung bình mẫu:', tb)
print('Độ lệch chuẩn mẫu:', dlc)
print('Sai số:', saiso)
print('Khoảng tin cậy:',
      (tb-saiso).round(3),
      (tb+saiso).round(3))
```

```python
#b
from scipy.stats import norm
import numpy as np

n = 100
so_cam_loai2 = 2 + 3

# Tỷ lệ mẫu
p_hat = so_cam_loai2 / n

# Độ tin cậy 90%
dotincay = 0.90
alpha = 1 - dotincay
z = norm.isf(alpha/2)

# Sai số
saiso = z * np.sqrt(p_hat * (1-p_hat) / n)

print('Tỷ lệ mẫu:', p_hat)
print('Sai số:', saiso.round(4))
print('Khoảng tin cậy:',
      (p_hat-saiso).round(3),
      (p_hat+saiso).round(3))
```
