# 2. Tỷ lệ đậu

Một môn học X được tổ chức mỗi học kỳ có tỷ lệ thí sinh học qua môn trung bình là 80%, và mỗi học kỳ có 250 sinh viên đăng ký dự thi. Tính giá trị kỳ vọng và độ lệch chuẩn của số lượng thí sinh không vượt qua môn học?

```python
#cau2
n = 250
p = 0.2

mean = n * p
std = np.sqrt(n * p * (1-p))

print("Kỳ vọng:", mean)
print("Độ lệch chuẩn:", std)
```

## 3. Tổng đài

Một tổng đài trung bình trong một giờ nhận được 180 cuộc gọi một cách ngẫu nhiên. Nếu trong 5 phút có từ hơn 20 cuộc gọi trở lên, thì tổng đài sẽ quá tải và người gọi đến sẽ phải chờ đợi. Giả sử số cuộc gọi nhận được trong 5 phút của tổng đài có phân phối Poisson. Tính:

a. Xác suất nhận được nhiều hơn 19 cuộc gọi mỗi 5 phút.

b. Trong 5 phút tổng đài có thể nhận được nhiều nhất bao nhiêu cuộc gọi để xác suất các cuộc gọi đều được xử lý ít nhất là 99%

```python
#cau3
#a. Xác suất nhận được hơn 19 cuộc gọi
from scipy.stats import poisson
lam = 15
p = 1 - poisson.cdf(19, lam)
print("Xác suất nhận được hơn 19 cuộc gọi",p)

#b. Nhiều nhất bao nhiêu cuộc gọi để xác suất xử lý được ≥ 99%?
k = poisson.ppf(0.99, 15)
print("Số cuộc gọi nhiều nhất để xác suất xử lý được ≥ 99%: ",k)
```

## 5. Độ dài chi tiết máy

Một công ty sản xuất làm ra các chi tiết máy với độ dài các chi tiết có phân phối chuẩn với kỳ vọng là 3000 mm và độ lệch chuẩn là 3 mm. Yêu cầu cho các chi tiết máy này là phải có chiều dài từ 2993 mm đến 3007 mm. Tính tỷ lệ sản phẩm lỗi của công ty

```python

mu = 3000
sigma = 3

p_dat = norm.cdf(3007, mu, sigma) - norm.cdf(2993, mu, sigma)
p_loi = 1 - p_dat

print("Tỷ lệ sản phẩm lỗi:", p_loi)
print("Tỷ lệ (%):", p_loi * 100, "%")
```

## 6. Mô phỏng phân phối Chuẩn

Giả sử điểm thi TOEIC là một biến ngẫu nhiên có phân phối chuẩn với kỳ vọng là 500, độ lệch chuẩn là 8. Bạn hãy:

a. Mô phỏng bằng cách phát sinh ngẫu nhiên điểm thi TOIEC của một nhóm gồm 50 người.

b. Tính điểm trung bình và độ lệch chuẩn về điểm của nhóm.

c. Tìm miền giá trị, và miền phân vị (IQR) của nhóm

d. Cho biết tỷ lệ đạt trên 450 điểm của nhóm. 

e. Vẽ đồ thị histogram của nhóm và so sánh với đồ thị hàm mật độ của phân phối chuẩn với kỳ vọng là 500 và độ lệch chuẩn là 8.

f. Thay vì mô phỏng 50 người, hãy mô phỏng nhóm với 100 người thi. Bạn có nhận xét gì?

```python
# a. Mô phỏng 50 người
np.random.seed(24)
TOEIC = np.random.normal(loc=500, scale=8, size=50)

# b. Trung bình và độ lệch chuẩn
mean_TOEIC = np.mean(TOEIC)
std_TOEIC = np.std(TOEIC)

print("Trung bình:", mean_TOEIC)
print("Độ lệch chuẩn:", std_TOEIC)

# c. Miền giá trị và IQR
print("Miền giá trị:", np.min(TOEIC), "-", np.max(TOEIC) )

Q1 = np.percentile(TOEIC, 25)
Q3 = np.percentile(TOEIC, 75)
print("IQR:", Q3 - Q1)

# d. Tỷ lệ đạt trên 450
ratio = np.mean(TOEIC > 450)
print("Tỷ lệ trên 450:", ratio)
print("Tỷ lệ (%):", ratio * 100, "%")

# e. Histogram và hàm mật độ chuẩn
x = np.linspace(470, 530, 1000)
y = norm.pdf(x, loc=500, scale=8)

plt.hist(TOEIC, bins=10, density=True, alpha=0.6, edgecolor='black')
plt.plot(x, y, linewidth=2)

plt.xlabel("Điểm TOEIC")
plt.ylabel("Mật độ")
plt.title("Histogram và hàm mật độ phân phối chuẩn")
plt.show()
# f. Mô phỏng 100 người
TOEIC_100 = np.random.normal(loc=500, scale=8, size=100)

print("Trung bình nhóm 100:", np.mean(TOEIC_100))
print("Độ lệch chuẩn nhóm 100:", np.std(TOEIC_100))

#Nhận xét f
print("Khi tăng từ 50 - 100 người, kết quả mô phỏng thường ổn định hơn, trung bình và độ lệch chuẩn của mẫu có xu hướng gần với 500 và 8 hơn. Histogram cũng thường có hình dạng gần phân phối chuẩn hơn.")
```
