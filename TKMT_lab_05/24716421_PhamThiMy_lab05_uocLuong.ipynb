```python
%matplotlib inline
import numpy as np
import math
from scipy import stats as stats
import matplotlib.pyplot as plt
```

### Bài 1: Tạo quần thể mô phỏng

Bạn hãy tạo một mẫu gồm 10000 phần tử mô phỏng chiều cao của nam thanh niên có giá trị từ 120cm - 200cm. Bạn hãy lưu kết quả vào biến **POP**.

Tính:
a. Chiều cao trung bình (kỳ vọng) của quần thể và độ lệch chuẩn về chiều cao của quần thể.

b. Tính tỷ lệ người cao trong quần thể, biết rằng thanh niên có chiều cao từ 180 trở lên được gọi là cao.

c. Vẽ histogram về chiều cao của quần thể. Theo bạn quần thể có phân phối chuẩn hay không?

d. Thử vẽ histogram và đồ thị hàm mật độ của phân phối chuẩn sử dụng tham số loc và scale bằng với kỳ vọng và độ lệch chuẩn của quần thể.

```python
POP = np.random.randint(120, 200, size=10000)
```

```python

#a. Chiều cao trung bình (kỳ vọng) của quần thể và độ lệch chuẩn về chiều cao của quần thể.
mean_of_POP = np.mean(POP)
print('Chiều cao trung bình của Quần Thể : {}'.format(mean_of_POP))

std_of_POP = np.std(POP)
print('Độ lệch chuẩn về chiều cao của quần thể : {}'.format(std_of_POP))

#b. Tính tỷ lệ người cao trong quần thể, biết rằng thanh niên có chiều cao từ 180 trở lên được gọi là cao.
tall = np.mean(POP >= 180)
print("Số người cao:", np.sum(POP >= 180))
print("Tỷ lệ người cao:", tall)
print("Tỷ lệ (%):", tall * 100, "%")


# c. Vẽ histogram về chiều cao của quần thể. Theo bạn quần thể có phân phối chuẩn hay không?
plt.figure(figsize=(10, 6))

plt.hist(
    POP,
    bins=30,
    edgecolor='black'
)
plt.xlabel("Chiều cao (cm)")
plt.ylabel("Tần số")
plt.title("Histogram chiều cao của quần thể")
plt.show()
# Histogram cho thấy chiều cao được phân bố khá đều trong khoảng từ 120 đến 199 cm
# Vì vậy, quần thể không có phân phối chuẩn.


# d. Thử vẽ histogram và đồ thị hàm mật độ của phân phối chuẩn sử dụng tham số loc và scale bằng với kỳ vọng và độ lệch chuẩn của quần thể.
plt.figure(figsize=(8, 5))

plt.hist(POP, bins=30, density=True, alpha=0.6, edgecolor='black')
x = np.linspace(120, 200, 1000)
y = stats.norm.pdf(x, loc=mean_of_POP, scale=std_of_POP)
plt.plot(x, y, linewidth=2)
plt.xlabel("Chiều cao (cm)")
plt.ylabel("Mật độ")
plt.title("Histogram và hàm mật độ phân phối chuẩn")

plt.show()

```

### Bài 4: Khảo sát phân phối của trung bình mẫu

a. Thực hiện bài 2 100 lần, mỗi lần bạn tính được trung bình mẫu. Vẽ đồ thị histogram cho 100 trung bình mẫu bạn tính được.

b. Thực hiện bài 3 100 lần, mỗi lần bạn tính được trung bình mẫu. Vẽ đồ thị histogram cho 100 trung bình mẫu bạn tính được.

c. Vẽ histogram của hai câu a, b trong cùng một hình để so sánh. Theo bạn hình dạng của histogram thay đổi nói lên điều gì?

Gợi ý: tạo một danh sách rỗng để chứa các trung bình mẫu tính được, sử dụng hàm append() để thêm giá trị trung bình vào danh sách sau mỗi lần tính.

```python
#a.Mẫu cỡ 20, lặp 100 lần
mean_sample_20 = []

for i in range(100):
    sample = np.random.choice(POP, size=20, replace=False)
    mean_sample_20.append(np.mean(sample))

plt.hist(mean_sample_20, bins=10, edgecolor='black')
plt.xlabel("Trung bình mẫu")
plt.ylabel("Tần số")
plt.title("Histogram của 100 trung bình mẫu (n=20)")
plt.show()

#b.Mẫu cỡ 100, lặp 100 lần
mean_sample_100 = []

for i in range(100):
    sample = np.random.choice(POP, size=100, replace=False)
    mean_sample_100.append(np.mean(sample))

plt.hist(mean_sample_100, bins=10, edgecolor='black')
plt.xlabel("Trung bình mẫu")
plt.ylabel("Tần số")
plt.title("Histogram của 100 trung bình mẫu (n=100)")
plt.show()

#c.Vẽ chung để so sánh
plt.hist(mean_sample_20, bins=10, alpha=0.6, label='n = 20')
plt.hist(mean_sample_100, bins=10, alpha=0.6, label='n = 100')

plt.xlabel("Trung bình mẫu")
plt.ylabel("Tần số")
plt.title("So sánh phân phối của trung bình mẫu")
plt.legend()
plt.show()
```

### Bài 6: Khảo sát phương sai mẫu

Bạn hãy mô phỏng 1000 lần lấy mẫu có kích cỡ 100. Vẽ histogram đại lượng phương sai của các mẫu (Phương sai bằng bình phương độ lệch chuẩn)

```python
#cau6
variance_100 = []

for i in range(1000):
    sample = np.random.choice(POP, size=100, replace=False)
    variance_100.append(np.var(sample))

plt.hist(variance_100, bins=15, edgecolor='black')

plt.xlabel("Phương sai mẫu")
plt.ylabel("Tần số")
plt.title("Histogram phương sai của 1000 mẫu (n=100)")
plt.show()
```

### Bài 7:  Bài Tập Ước Lượng 

Khảo sát cân nặng (kg) của gà khi xuất chuồng, người ta cân một số con và có kết quả như sau:

> 2.1, 1.8, 2.0, 2.3, 1.7, 1.5, 2.0, 2.2, 1.8

Giả sử cân nặng của gà là biến ngẫu nhiên có phân phối chuẩn. Hãy lập khoảng tin cậy cho cân nặng trung bình của  khi xuất chuồng với độ tin cậy là 95%. Trong 2 trường hợp:

a, Biết $\sigma = 0.3$

b, Không biết $\sigma$


```python
data = np.array([2.1, 1.8, 2.0, 2.3, 1.7, 1.5, 2.0, 2.2, 1.8])

n = len(data)
mean = np.mean(data)

print("Trung bình mẫu:", mean)
#a. Biết σ = 0.3
#Biết độ lệch chuẩn quần thể → dùng Z:

sigma = 0.3
confidence = 0.95

z = stats.norm.ppf(1 - (1-confidence)/2)

error = z * sigma / np.sqrt(n)

lower = mean - error
upper = mean + error

print("Khoảng tin cậy 95%:", (lower, upper))

#b. Không biết σ
#Không biết độ lệch chuẩn quần thể → dùng t-Student:

s = np.std(data, ddof=1)

t = stats.t.ppf(1 - (1-confidence)/2, n-1)

error = t * s / np.sqrt(n)

lower = mean - error
upper = mean + error

print("Độ lệch chuẩn mẫu:", s)
print("Khoảng tin cậy 95%:", (lower, upper))
```
